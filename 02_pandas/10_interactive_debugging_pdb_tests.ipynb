# interactive debugging pdb — exercises

    13 tasks that test what the cheat sheet `10_interactive_debugging_pdb.ipynb` covers. Try each one
    **before** looking at the cheat sheet.

    How it works:
    1. Read the task. Write your code in the cell below it, assigning the result to the
       variable the task names (usually `answer`).
    2. Run the cell: the last line `qN.check()` tells you ✅ or ❌ with a short reason.
    3. Stuck? Uncomment `qN.hint()` for a nudge, or `qN.solution()` to see the reference code.

Tip: after a failing cell you can type `%debug` in a new cell and use `w`, `p`, `u`,
`d`, `q`. The setup of each task captures the traceback in `tb` (and the exception in
`exc`) so the tasks also work without an interactive prompt: `_frames(tb)` returns
`[(function_name, locals), ...]`, innermost frame last.

```python
import sys; sys.path.append("..")
from quantlearn import load_questions
q1, q2, q3, q4, q5, q6, q7, q8, q9, q10, q11, q12, q13 = load_questions("02_pandas/10_interactive_debugging_pdb")

import sys
import traceback
import numpy as np
import pandas as pd

def _frames(tb):
    # helper: list of (function name, locals dict) from a traceback, innermost last
    out = []
    while tb is not None:
        out.append((tb.tb_frame.f_code.co_name, tb.tb_frame.f_locals))
        tb = tb.tb_next
    return out
```
